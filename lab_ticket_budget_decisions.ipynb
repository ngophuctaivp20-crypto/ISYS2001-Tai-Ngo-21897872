Design a program that acts as an expense classifier, sorting a user's spending into categories (Housing, Food, Transport, Debts & Extras, and Other) and reacting to each entry as High, Medium, or Low based on the amount spent. The program should ask the user to input an expense category and amount, then use if, elif, else, and, and or statements to determine both the category and spending level.

```python
def classify_expense(category, amount):
    category = category.lower()

    # Sort into categories using if, elif, else, and, or
    if ("house" in category or "rent" in category or "mortgage" in category
            or "electric" in category or "water" in category or "gas bill" in category
            or "internet" in category or "wifi" in category or "insurance" in category
            or "property tax" in category or "maintenance" in category
            or "trash" in category or "hoa" in category):
        group = "Housing"
        low, high = 500, 1500

    elif ("food" in category or "grocery" in category or "groceries" in category
            or "restaurant" in category or "dining" in category or "takeout" in category
            or "coffee" in category or "snack" in category or "supermarket" in category):
        group = "Food"
        low, high = 100, 400

    elif ("transport" in category or "gas" in category or "fuel" in category
            or "uber" in category or "taxi" in category or "bus" in category
            or "train" in category or "car insurance" in category
            or "parking" in category or "car repair" in category):
        group = "Transport"
        low, high = 50, 200

    elif ("debt" in category or "loan" in category or "credit card" in category
            or "installment" in category or "interest" in category):
        group = "Debts & Extras"
        low, high = 50, 300

    elif ("subscription" in category or "entertainment" in category or "shopping" in category
            or "netflix" in category or "spotify" in category or "gym" in category
            or "clothes" in category or "gift" in category or "travel" in category
            or "misc" in category):
        group = "Other"
        low, high = 20, 100

    else:
        group = "Other"
        low, high = 20, 100

    # React based on amount using if, elif, else, and, or
    if amount >= high:
        level = "High"
    elif amount <= low and amount > 0:
        level = "Low"
    elif amount > low and amount < high:
        level = "Medium"
    else:
        level = "Invalid"

    return group, level


# --- Interactive input loop ---
print("Expense Classifier — type 'done' when finished\n")

totals = {}  # tracks total spent per category

while True:
    category_input = input("Enter expense category (e.g. rent, electricity, groceries...: ")

    if category_input.lower() == "done":
        print("\nFinished classifying expenses.")
        break

    # Reject numeric input for category
    if category_input.replace(".", "", 1).isdigit():
        print("⚠️ That looks like a number, not a category name. Please enter a category like 'rent' or 'food'.\n")
        continue

    amount_input = input("Enter amount ($): ")

    try:
        amount_input = float(amount_input)
    except ValueError:
        print("⚠️ Please enter a valid number for the amount.\n")
        continue

    group, level = classify_expense(category_input, amount_input)
    print(f"→ Category: {group:<15} Level: {level}\n")

    # Update running totals
    totals[group] = totals.get(group, 0) + amount_input

# --- Summary ---
if totals:
    print("Summary of spending by category:")
    grand_total = 0
    for group, total in totals.items():
        print(f"  {group:<15} ${total:.2f}")
        grand_total += total
    print(f"\nGrand Total: ${grand_total:.2f}")
else:
    print("No expenses were entered.")
```
